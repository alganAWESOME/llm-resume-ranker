```python
import anthropic
```

```python
def write_about_animal(animal='dog'):
    client = anthropic.Anthropic()

    message = client.messages.create(
        model="claude-3-haiku-20240307",
        max_tokens=1996,
        temperature=0,
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": f"Write a very short sentence about {animal}s."
                    }
                ]
            }
        ]
    )

    return message
```

```python
import concurrent

def my_func(animals=['cat', 'dog', 'mouse', 'rabbit']):

    with concurrent.futures.ThreadPoolExecutor(max_workers=100) as executor:
            # Submit tasks to the executor
            futures = [executor.submit(write_about_animal, animal) for animal in animals]

            # Collect the results as they complete
            for future in concurrent.futures.as_completed(futures):
                print(future.result())
```
